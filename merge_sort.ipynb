```python
# helper function to merge_sort
def final_merge(left_arr,right_arr,arr):

    i = j = k =0  # indices for left, right, and arr

    while i < len(left_arr) and j < len(right_arr):
        if left_arr[i] <= right_arr[j]:
            arr[k] = left_arr[i]
            i += 1
        else:
            arr[k] = right_arr[j]
            j += 1
        k += 1
        
    # Copy remaining elements
    while i < len(left_arr):
        arr[k] = left_arr[i]
        i += 1
        k += 1
    while j < len(right_arr):
        arr[k] = right_arr[j]
        j += 1
        k += 1
        
    return
```

```python
# Recursive merge sort
def merge_sort(arr):
    if not arr or len(arr) == 1: # base case
        return arr

    mid_index = len(arr)//2

    left_arr = arr[:mid_index]
    right_arr = arr[mid_index:]

    merge_sort(left_arr)
    merge_sort(right_arr)

    final_merge(left_arr,right_arr,arr)

    return arr
```

```python
# Example
arr=[4,3,2,5,1,6,7,0,7,2,1,0]
res = merge_sort(arr)
print(res)
```
